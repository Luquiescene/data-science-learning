# 一、ndarray
## 1.ndarray 的特性
### 多维性

```python
import numpy as np
```

```python
arr = np.array(5)  # 创建0维ndarray数组
print(arr)
print('arr的维度是:', arr.ndim)  # 查看数组维度 number of dimensions
```

```python
arr = np.array([1,2,3])  # 创建1维ndarray数组
print(arr)
print('arr的维度是:', arr.ndim)  # 查看数组维度 number of dimensions
```

```python
arr = np.array([[1,2,3], [4,5,6]])  # 创建2维ndarray数组
print(arr)
print('arr的维度是:', arr.ndim)  # 查看数组维度 number of dimensions
```

### 同质性

```python
arr = np.array([1,'hello'])  # 不同类型的元素会被强制转换为同一类型
print(arr)
```

```python
arr = np.array([1, 2.5])  # 不同类型的元素会被强制转换为同一类型
print(arr)
```

## 2. ndarray的属性

```python
arr = np.array(1)
print(arr)
print('数组的形状：', arr.shape)  # 查看数组形状 shape
print('数组的维度：', arr.ndim)  # 查看数组维度 number of dimensions
print('元素的个数：', arr.size)  # 查看数组元素个数 size
print('数组的类型：', arr.dtype)  # 查看数组元素类型 dtype
print('元素的转置：', arr.T)
```

```python
arr = np.array([1, 2.5, 3])
print(arr)
print('数组的形状：', arr.shape)  # 查看数组形状 shape
print('数组的维度：', arr.ndim)  # 查看数组维度 number of dimensions
print('元素的个数：', arr.size)  # 查看数组元素个数 size
print('数组的类型：', arr.dtype)  # 查看数组元素类型 dtype
print('元素的转置：', arr.T)
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr)
print('数组的形状：', arr.shape)  # 查看数组形状 shape
print('数组的维度：', arr.ndim)  # 查看数组维度 number of dimensions
print('元素的个数：', arr.size)  # 查看数组元素个数 size
print('数组的类型：', arr.dtype)  # 查看数组元素类型 dtype
print('元素的转置：', arr.T)
```

## 3. ndarray 的创建

```python
# 基础的创建方法
list1 = [4, 5, 6]
arr = np.array(list1, dtype=np.float64)  # 指定数据类型
print(arr.ndim)  # 属性和方法的区分：是否带括号，带括号的是方法，不带括号的是属性
print(arr)
```

```python
# copy
arr1 = np.copy(arr)  # 元素跟原始的数组相同，但是不是一个数组了
print(arr1)
arr1[0] = 8
print(arr1)
print(arr)  # 原始数组不变
```

```python
# 预定义形状
# 全0 全1 未初始化 固定值
# 全0
arr = np.zeros((2,3), dtype = int)  # 创建一个2行3列的全0数组
print(arr)
print(arr.dtype)
```

```python
arr = np.zeros((200,), dtype = int)  # 创建一个200行1列的全0数组
print(arr)
```

```python
# 全1
arr = np.ones((5,8), dtype = int)  # 创建一个5行8列的全1数组
print(arr)
```

```python
# 未初始化
arr = np.empty((4,3))  # 创建一个4行3列的未初始化数组，执行效率快
print(arr)
```

```python
# 固定值
arr = np.full((3,4), 2026)  # 创建一个3行4列的全2026数组
print(arr)
```

```python
arr1 = np.zeros_like(arr)  # 创建一个与arr形状相同的全0数组
print(arr1)
arr1 = np.empty_like(arr)  # 创建一个与arr形状相同的未初始化数组
print(arr1)
arr1 = np.ones_like(arr)  # 创建一个与arr形状相同的全1数组
print(arr1)
arr1 = np.full_like(arr, 2025)  # 创建一个与arr形状相同的全2025数组
print(arr1)
```

```python
# 等差数列 2 4 6 8 10
arr = np.arange(2, 12, 2) # start, stop(不取), step
print(arr)
```

```python
# 等间隔数列
arr = np.linspace(0, 100, 5)  # start, stop(取), num
print(arr)
```

```python
arr = np.linspace(0, 100, 5, dtype=int)
print(arr)
arr = np.arange(0, 101, 25, dtype=int)
print(arr)
```
