```python
import pandas as pd
import numpy as np
```

```python
# DataFrame 的创建方式
# 通过Series创建
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([6,7,8,9,10])
df = pd.DataFrame({'第一列':s1,'第二列':s2})
print(type(df['第一列'])) # 每一列是一个series

# 通过字典创建
df = pd.DataFrame(
    {
        'name':['tom','jack','alice','bob','allen'],
        'age':[15,17,20,26,30],
        'score':[60.5,80,30.6,70,83.5]
    },index=[1,2,3,4,5],columns=['name','score','age']
)
print(df)
```

```python
# dataframe 的属性
print('行索引：', df.index)
print('列标签：', df.columns)
print('值：', df.values)

print('维度：', df.ndim)
print('形状：', df.shape)
print('元素个数：', df.size)
print('数据类型：\n', df.dtypes)
```

```python
# 行列转置
print(df.T)
print(df.T.index)
```

```python
# 获取元素 loc iloc at iat
# 某行
print(df.loc[4])
print(df.iloc[3])

# 某列
print(df.loc[:,'name'])
print(df.iloc[:,0])

# 单个元素
print(df.at[3,'score'])
print(df.iat[2,1])
print(df.loc[3,'score'])
print(df.iloc[2,1])
```

```python
# 获取单列数据
print(df['name'])
print(type(df['name']))
print(df.name)
print(type(df.name))
print(df[['name']])
print(type(df[['name']]))
```

```python
# 获取多列数据
print(df[['name','score']])
```

```python
# 查看部分数据
print(df.head(2))
print(df.tail(3))
```

```python
# 使用布尔索引筛选数据
print(df[df.score>70])
print(df[(df.score>70) & (df.age<20)])
```

```python
# 随机抽样
df.sample(3)
```
