```python
import numpy as np
import pandas as pd
```

```python
# 1. 学生成绩统计
# 生成随机成绩
np.random.seed(42)
scores = pd.Series(np.random.randint(50,101,10),index=['学生'+str(i) for i in range(1,11)])
print(scores)
```

```python
# 我的代码
print('平均分',scores.mean())
print('最高分',scores.max())
print('最低分',scores.min())
print(len(scores[scores>scores.mean()]))
print(scores[scores>scores.mean()].count())
```

```python
# 2. 温度数据分析
temperatures = pd.Series([28,31,29,32,30,27,33],index=['周一','周二','周三','周四','周五','周六','周日'])
# 找出温度超过30的天数
print(temperatures[temperatures>30].count())
print(temperatures.mean())
print(temperatures.sort_values(ascending=False)) # 降序

# 计算温度相差最多的2天
# 28 31 29 32 30 27 33
# none 3 -2 3 -2 -3 6
t3 = temperatures.diff().abs() # 计算Series的变化值的绝对值

print('温度变化最大的2天',t3.sort_values(ascending=False).index[:2].tolist())
```

```python
# 3. 股票价格分析
# 日期序列
date = pd.date_range('2025-06-01',periods=6)
print(list(date))

prices = pd.Series([102.3,103.5,105.1,104.8,106.2,107.0,106.5,108.1,109.3,110.2],index=pd.date_range('2023-01-01',periods=10))
print(prices)
```

```python
# 计算每日收益率  当日收盘价/前日收盘价 - 1
a = prices.pct_change()
print(a)
```

```python
# 收益率最高的日期
print(a.idxmax())
# 收益率最低的日期
print(a.idxmin())
```

```python
# 波动率
print(a.std())
```

```python
# 4. 销售数据分析
sales = pd.Series([120,135,145,160,155,170,180,175,190,200,210,220],index=pd.date_range('2022-01-01',periods=12,freq='MS'))
print(sales)
```

```python
# 季度的平均销量
print(sales.resample('QS').mean()) # 重新采样
```

```python
# 销量最高的月份
print(sales.idxmax())
```

```python
# 月环比增长率
print(sales.pct_change())
```

```python
# 连续增长超过2个月的月份
# 滑动窗口算法
a = sales.pct_change()
b = a > 0
print(list(b[b.rolling(3).sum() == 3].keys()))
```

```python
# 5. 每小时销售数据分析
np.random.seed(42)
hourly_sales = pd.Series(np.random.randint(1,100,24),index=pd.date_range('2025-01-01',periods=24,freq='h'))
print(hourly_sales)
```

```python
# 按天重采样计算每日总销售额
day_sales = hourly_sales.resample('D').sum()
print(day_sales)
```

```python
# 计算每天营业时间和非营业时间的销售额比例
# print(hourly_sales.between_time('8:00','22:00')) # 筛选一段时间内的Series

business_hours_sales = hourly_sales[(hourly_sales.index.hour>=8) & (hourly_sales.index.hour<=22)]
not_business_hours_sales = hourly_sales.drop(business_hours_sales.index)
print(business_hours_sales.sum() / not_business_hours_sales.sum())

# 找出销售额最高的3个小时
print(list(hourly_sales.nlargest(3).keys()))
```
