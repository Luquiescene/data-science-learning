```python
# series 的创建
import pandas as pd
s = pd.Series([1,2,3,4,5])
print(s)

# 自定义索引
s = pd.Series([1,2,3,4,5], index=['A','B','C','D','E'])
print(s)

# 定义name
s = pd.Series([1,2,3,4,5], index=['A','B','C','D','E'], name='月份')
print(s)
```

```python
# 通过字典来创建
s = pd.Series({'a':1,'b':2,'c':3,'d':4,'e':5})
print(s)

s1 = pd.Series(s,index=['a','c'])
print(s1)
```

```python
# series 的属性

print(s)
print(s.index)
print(s.values)
print(s.shape, s.ndim, s.size)
s.name = 'test'
print(s.dtype, s.name)

print(s.loc['a'])  # 显式索引
print(s.iloc[0])  # 隐式索引
print(s.at['a'])  # 不支持切片
print(s.iat[0])
```

```python
# 访问数据
# print(s[0]) # 不推荐使用，因为索引不清楚是显式还是隐式

print(s.iloc[0])
print(s)
print(s[s<3]) # 布尔索引
s['f'] = 6
print(s.head()) # 返回series的前5行
print(s.tail()) # 返回series的后5行
print(s.head(2))
```

```python
# 常见函数
import numpy as np
s = pd.Series([10,2,np.nan,None,3,4,5], index=['A','B','C','D','E','F','G'], name='data')
print(s)
```

```python
s.head() # 默认取前5行的数据
s.tail() # 默认取后5行的数据
```

```python
# 查看所有的描述性信息
s.describe()
```

```python
# 获取元素个数(忽略缺失值)
print(s.count())
```

```python
# 获取索引
print(s.keys()) # 方法
print(s.index) # 属性
```

```python
print(s.isna()) # 检查Series里的每一个元素是否为缺失值
```

```python
s.isin([4,5,6]) # 检查是否有4,5,6在Series中
```

```python
print(s.mean()) # 平均数
print(s.sum()) # 总和
print(s.std()) # 标准差
print(s.var()) # 方差
print(s.min()) #  最小值
print(s.max()) # 最大值
print(s.median()) # 中位数
```

```python
print(s.quantile(0.25)) # 分位数
```

```python
# 众数
s['H'] = 4
print(s.mode())
```

```python
print(s.value_counts()) # 每个元素的频数
```

```python
print(s.drop_duplicates()) # 去重，缺失值也去重，返回Series
print(s.unique()) # 去重，返回列表
print(s.nunique()) # 去重后的元素个数
```

```python
# 排序 值、索引
print(s.sort_index()) # 按索引排序
print(s.sort_values()) # 按值排序
```
